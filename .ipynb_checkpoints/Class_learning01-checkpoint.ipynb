```python
class student:
    def __init__(self,name,marks):
        self.name = name
        self.marks = marks

    def total(self):
        total_marks = 0
        for i in range(len(self.marks)):
            total_marks = total_marks + self.marks[i]
        return total_marks

    def average(self,total_marks):
        avg = total_marks/len(self.marks)
        return avg

    def maximum(self):
        max_mark = max(self.marks)
        return max_mark

    def print_result(self,total_marks,average,maximum):
        print("Total marks scored by " + self.name + " is " + str(total_marks))
        print("Average marks scored by " + self.name + " is " + str(average))
        print("Maximum marks scored by " + self.name + " is " + str(maximum))
    
s1=student("Arshad",[90,89,99,95,100])
total_marks = s1.total()
average = s1.average(total_marks)
maximum = s1.maximum()
s1.print_result(total_marks,average,maximum)
```
