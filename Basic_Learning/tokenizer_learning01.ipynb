```python
import re
```

```python
class text_tokenizer:
    def __init__(self,vocab):
        self.str_to_int = vocab
        self.int_to_str = {i:s for s,i in vocab.items()}

    def encode(self, text):
        pre_processed = re.split(r'[(.,!)]|\s', text)
        pre_processed = [i.strip() for i in pre_processed if i.strip()]
        ids = []
        for word in pre_processed:
            if word in self.str_to_int:
                ids.append(self.str_to_int[word])
        return ids

    def decode(self,ids):
        word = []
        for i in ids:
            if i in self.int_to_str:
                word.append(self.int_to_str[i])
                string = " ".join(word)
        return string     
```

```python
raw_text = "Arshad is learning  python! and machine learning. He is creating a LLM from scratch"
vocab = re.split(r'[(.,!)]|\s',raw_text)
final_vocab = []
for word in vocab:
    if word.strip():
        final_vocab.append(word)
    
dic_vocab = {}
for integer,token in enumerate(final_vocab):
    dic_vocab[token] = integer

```
