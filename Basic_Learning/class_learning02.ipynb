```python
class BankAccount:
    def __init__(self,name,balance,transaction):
        self.name = name
        self.balance = balance
        self.transaction = transaction

    def deposit(self,amount):
        if amount<0:
            print("Incorrect amount")
        else:
            self.balance += amount
            self.transaction.append("Deposited amount " + str(amount))  

    def withdraw(self,amount):
        if self.balance<amount:
            print("Insufficient Balance")
        else:
            self.balance -= amount
            self.transaction.append("Withdrew amount " + str(amount))

    def check_balance(self):
        print("Your current balance is " + str(self.balance))
    
   
    def all_transaction(self):
        for trans in self.transaction:
            print(trans)
        
    
    


B1 = BankAccount("Arshad",0,[])
B1.deposit(1000)
B1.withdraw(750)
B1.deposit(250)
B1.withdraw(100)
B1.all_transaction()
B1.check_balance()
```
