```python
raw_text = "This is the second coding to make a tokeniser with minimal outside resource. Insha Allah, I'll succeed!"
```

```python
vocab = {}
for idx,word in enumerate(pre_process):
    if word.strip():
        vocab[idx]=word.strip()
        
```

```python
vocab={idx:word for idx,word in enumerate(pre_process) if word.strip()}
rev_vocab = {word:idx for idx,word in vocab.items()}
```

```python
class Tokeniser_v2:
    def __init__(self,vocab):
        self.int_to_str = vocab
        self._str_to_int = {word:idx for idx,word in vocab.items()}

    def encode(self,text):
        preprocess = re.split(r'[,.()!"]|\s',text)
        inp_words = [ word.strip() for word in preprocess if word.strip()]
        ids=[]

        for w in inp_words:
            if w in self._str_to_int:
                ids.append(self._str_to_int[w])
            else:
                ids.append("|<unk>|")
        return ids

    def decode(self,ids):
        char = []
        for i in ids:
            if i in self.int_to_str:
                char.append(self.int_to_str[i])
                string = " ".join(char)              
        return string
    
```

```python
t1 = Tokeniser_v2(vocab)
inp_text = "second aka coding to make a tokeniser"
ids = t1.encode(inp_text)
print(ids)
t1.decode(t1.encode(inp_text))
```
